```python
import pandas as pd

roster = pd.read_csv("../data/lka/sri_lanka_features19_updated.csv")#, nrows=500)

feat_names = pd.read_csv("../data/lka/features_explanations.csv")
```

```python
feat_names_av = feat_names['codename']
```

```python
feat_set = set(feat_names_av)
in_roster_not_in_feat_names_av = [col for col in roster.columns if col not in feat_set]
all_cols_with_flag = pd.DataFrame({"value": roster.columns})
all_cols_with_flag["in_feat_names_av"] = all_cols_with_flag["value"].isin(feat_set)
print(all_cols_with_flag.to_string(index=False))
print(in_roster_not_in_feat_names_av)
```
